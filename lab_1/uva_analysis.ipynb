```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
import ipywidgets as widgets
```

```python
df = pd.read_csv ('ForeignGifts_edu.csv')
```

# UVA Findings

```python
uva_df = df[df['Institution Name'] == 'University of Virginia']
```

```python
#Count of foreign gifts to UVA
print(uva_df.shape[0])

#Total amount of foreign gifts to UVA
print(uva_df['Foreign Gift Amount'].sum())

#Average amount of foreign gifts to UVA
print(uva_df['Foreign Gift Amount'].mean())
```

The output of the code above shows that UVA has recieved 98 total foreign gifts totaling to $22,189,238. The average amount per foreign gift for UVA was about $226,421

The above output shows that the top gift type for UVA in terms of amount of gifting was contracts which provided almost double the dollar amount of gifting compated to monetary gifts.

```python
#Histogram of UVA foreign gift amount
gift_amount_hist = px.histogram(uva_df, x='Foreign Gift Amount', title='Distribution of Amount per Foreign Gift at UVA', labels={'Foreign Gift Amount': 'Foreign Gift Amount ($)'})
gift_amount_hist.update_layout(yaxis_title='Count of Foreign Gifts')

gift_amount_hist.show()
```

When looking at the distribution of foreign gift amounts to UVA, the graph above shows that the majority of foreign gifts that UVA recieves are less than $200,000.

```python
#Histogram of country of giftor to UVA
uva_country_hist = px.histogram(uva_df, x='Country of Giftor', 
                                title='Count of Foreign Gifts by Country',
                                labels={'Country of Giftor': 'Country'})
uva_country_hist.update_layout(yaxis_title='Count of Foreign Gifts')

#Bar of gift amount to UVA by country
uva_gift_amount_country = uva_df.groupby('Country of Giftor', as_index=False).sum("Foreign Gift Amount")
uva_gift_amount_bar = px.bar(uva_gift_amount_country, x='Country of Giftor', 
                             y='Foreign Gift Amount', 
                             title='Total Foreign Gift Amount by Country', 
                             labels={'Foreign Gift Amount': 'Total Gift Amount ($)', 'Country of Giftor': 'Country'})

uva_country_hist.show()
uva_gift_amount_bar.show()
```

Breaking this down by where these gifts are coming from, the graph titled 'Count of Foreign Gifts by Country' shows that UVA recieves the most gifts by far from England. But it can be seen in the second graph titled 'Total Foreign Gift Amount by Country', that Bangledesh and Sweeden have given UVA the most foreign gifting in dollar amount. These both have a total gifting amount that is at least double that of England.

```python
#Breakdown of number of gifts by gift type
pd.DataFrame(uva_df['Gift Type'].value_counts().reset_index(name = 'Count of Foreign Gifts'))
```

I next wanted to see what type of foreign gifts UVA was recieving at the highest volume. The table above shows that the largest number of foriegn gifts to UVA are contracts and the only other type of gift recieved is monetary gifts. 

```python
uva_df.groupby("Gift Type").sum("Foreign Gift Amount").sort_values(by="Foreign Gift Amount", ascending=False)[['Foreign Gift Amount']]
```

When looking at the gift amounts, UVA recieved a higher dollar amount in foreign gifting through contracts at $14,807,636 than through monetary gifting which only earned them $7,381,602.

# Comparison to Other Top Public Universities

```python
#Creating data set for UVA and peer institutions
top_public_universities = ['University of Virginia', 'University of Michigan - Ann Arbor', 'University of North Carolina - Chapel Hill', 'University of California, Berkeley', 'University of California, Los Angeles', 'University of California, San Diego', 'University of Florida', 'University of Texas at Austin']
uva_and_peers_df = df[df['Institution Name'].isin(top_public_universities)]
```

To be able to put UVA's performance in foreign gifting into context I wanted to compare it to the performance of other top public universities. I chose 7 other universities to compare UVA to: 
- University of Michigan - Ann Arbor
- University of North Carolina - Chapel Hill
- University of California, Berkeley
- University of California, Los Angeles
- University of California, San Diego
- University of Florida, University of Texas at Austin.

```python
#Number of foreign gifts by institution
uva_and_peers_df.groupby("Institution Name").sum("Foreign Gift Amount").sort_values(by="Foreign Gift Amount", ascending=False)[['Foreign Gift Amount']]
```

First comparing UVA's $22,189,238 total foreign gifting to other the other top public universities I selected, it ranks last. The top of these public universties in terms of total foreign gift amount is Berkely at $2,94,479,904 which is over 13 times that of UVA. University of Michigan - Ann Arbor comes in second at $287,336,783 which is still 12 times that of UVA.

```python
pd.DataFrame(uva_and_peers_df['Institution Name'].value_counts().reset_index(name = 'Count of Foreign Gifts'))
```

There is a similar story when looking at the total number of foreign gifts recieved by each institution where UVA is second-to-last followed only by University of Florida. University of California, Los Angeles recieved the largest number of foreign gifts at 3,916 and University of Michigan - Ann Arbor	came in second at 715.

```python
uvapeers_grouped = uva_and_peers_df.groupby('Country of ', as_index=False).sum("Foreign Gift Amount")
uvapeers_bar = px.bar(uvapeers_grouped, x='Institution Name', 
                        y='Foreign Gift Amount', 
                        title='Total Foreign Gift Amount by Institution', 
                        labels={'Foreign Gift Amount': 'Total Gift Amount ($)', 'Institution Name': 'Academic Institution'})


uvapeers_bar.show()
```

```python
uvapeers_grouped  = uva_and_peers_df.groupby('Country of Giftor', as_index=False).sum("Foreign Gift Amount").sort_values(by='Foreign Gift Amount', ascending=False).head(20)
uvapeers_grouped_bar = px.bar(uvapeers_grouped, x='Country of Giftor', 
                             y='Foreign Gift Amount', 
                             title='Total Foreign Gift Amount by Country', 
                             labels={'Foreign Gift Amount': 'Total Gift Amount ($)', 'Country of Giftor': 'Country'})

uvapeers_grouped_bar.show()
```

I also wanted to see what countries give the most to the top public universities and the graph above shows that England provided the highest dollar amount of gifting to these universities.

Because these contracts appear to provide a higher amount of total gifting, I also looked at a higher level accross all academic instiutions to see if contracts make up a higher proportion of total gifting overall as well. In the second table below, it shows that contracts make up over double the total foreign gifting than monetary contracts. This might indicate that it could be worth exploring further contract opporunities within England, which UVA already has a strong monetary gifting relationship with, and somewhat of contract giving history as well.
